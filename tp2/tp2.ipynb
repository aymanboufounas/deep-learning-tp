```python
from tensorflow import keras 
from keras.datasets import boston_housing
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import  Dense , Input
from tensorflow.keras.optimizers import Adam , SGD , RMSprop
from sklearn.preprocessing import StandardScaler
```

```python
(x_train, y_train), (x_test, y_test) = boston_housing.load_data()
```

```python
df = pd.DataFrame(x_train)
df["price"] = y_train
```

```python
plt.hist(y_train , bins=20)
plt.xlabel("price")
plt.ylabel("number of house")
```

```python
plt.boxplot(y_train)
```

```python
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
model = Sequential(
[    Dense(
        64 , activation='relu' , input_shape = (13,)
    ),
    Dense(
        32 , activation='relu'
    ),
    Dense(1)]
)
```

```python
model.summary()
```

```python
model_sgd = Sequential([
    Input(shape=(13,)),
    Dense(64, activation='relu'),
    Dense(32, activation='relu'),
    Dense(1)
])
```

```python
model_sgd.compile (
    optimizer = SGD(learning_rate=0.01),
    loss='mse',
    metrics=['mae']
)
```

```python
train_sgd = model_sgd.fit(
    x_train,
    y_train,
    epochs=50,
    batch_size=32,
    validation_split=0.2

)
```

```python
model_momentum = Sequential([
    Input(shape=(13,)),
    Dense(64, activation='relu'),
    Dense(32, activation='relu'),
    Dense(1)
])
```

```python
model_momentum.compile(
    optimizer=SGD(learning_rate=0.01, momentum=0.9),
    loss='mse',
    metrics=['mae']
)

train_momentum = model_momentum.fit(
    x_train, y_train,
    epochs=50,
    batch_size=32,
    validation_split=0.2
)
```

```python
model_rmsprop = Sequential([
    Input(shape=(13,)),
    Dense(64, activation='relu'),
    Dense(32, activation='relu'),
    Dense(1)
])

model_rmsprop.compile(
    optimizer=RMSprop(learning_rate=0.001),
    loss='mse',
    metrics=['mae']
)

train_rmsprop = model_rmsprop.fit(
    x_train, y_train,
    epochs=50,
    batch_size=32,
    validation_split=0.2
)
```

```python
model_adam = Sequential([
    Input(shape=(13,)),
    Dense(64, activation='relu'),
    Dense(32, activation='relu'),
    Dense(1)
])

model_adam.compile(
    optimizer=Adam(learning_rate=0.001),
    loss='mse',
    metrics=['mae']
)

train_adam = model_adam.fit(
    x_train, y_train,
    epochs=50,
    batch_size=32,
    validation_split=0.2
)
```

```python
plt.plot(train_sgd.history['mae'], label='SGD')
plt.plot(train_momentum.history['mae'], label='Momentum')
plt.plot(train_rmsprop.history['mae'], label='RMSprop')
plt.plot(train_adam.history['mae'], label='Adam')
plt.xlabel('Epochs')
plt.ylabel('MAE')
plt.title('Boston Housing - Train MAE')
plt.legend()
plt.show()
```

```python
plt.plot(train_sgd.history['val_mae'], label='SGD')
plt.plot(train_momentum.history['val_mae'], label='Momentum')
plt.plot(train_rmsprop.history['val_mae'], label='RMSprop')
plt.plot(train_adam.history['val_mae'], label='Adam')
plt.xlabel('Epochs')
plt.ylabel('MAE')
plt.title('Boston Housing - Validation MAE')
plt.legend()
plt.show()
```

```python
scores = {
    'SGD': model_sgd.evaluate(x_test, y_test, verbose=0)[1],
    'Momentum': model_momentum.evaluate(x_test, y_test, verbose=0)[1],
    'RMSprop': model_rmsprop.evaluate(x_test, y_test, verbose=0)[1],
    'Adam': model_adam.evaluate(x_test, y_test, verbose=0)[1]
}
print(scores)

v10 = {
    'SGD': train_sgd.history['val_mae'][9],
    'Momentum': train_momentum.history['val_mae'][9],
    'RMSprop': train_rmsprop.history['val_mae'][9],
    'Adam': train_adam.history['val_mae'][9]
}
print('a. Meilleure MAE après 10 epochs :', min(v10, key=v10.get))
print('b. Plus petite erreur sur test :', min(scores, key=scores.get))
```

```python
model_sgd_01 = Sequential([
    Input(shape=(13,)),
    Dense(64, activation='relu'),
    Dense(32, activation='relu'),
    Dense(1)
])

model_sgd_01.compile(
    optimizer=SGD(learning_rate=0.1),
    loss='mse',
    metrics=['mae']
)

train_sgd_01 = model_sgd_01.fit(
    x_train, y_train,
    epochs=50,
    batch_size=32,
    validation_split=0.2,
    callbacks=[keras.callbacks.TerminateOnNaN()]
)
```

```python
plt.plot(train_sgd.history['val_mae'], label='SGD 0.01')
plt.plot(train_sgd_01.history['val_mae'], label='SGD 0.1')
plt.xlabel('Epochs')
plt.ylabel('MAE')
plt.title('Boston - Learning rate')
plt.legend()
plt.show()

print('c. Un learning rate élevé peut accélérer, mais aussi rendre l’entraînement instable.')
print('d. Adam adapte le learning rate à chaque poids, donc il est souvent plus stable que SGD.')
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
model_sgd_ca = Sequential([
    Input(shape=(8,)),
    Dense(64, activation='relu'),
    Dense(32, activation='relu'),
    Dense(1)
])

model_sgd_ca.compile(
    optimizer=SGD(learning_rate=0.01),
    loss='mse',
    metrics=['mae']
)

train_sgd_ca = model_sgd_ca.fit(
    x_train, y_train,
    epochs=50,
    batch_size=32,
    validation_split=0.2
)
```

```python
model_momentum_ca = Sequential([
    Input(shape=(8,)),
    Dense(64, activation='relu'),
    Dense(32, activation='relu'),
    Dense(1)
])

model_momentum_ca.compile(
    optimizer=SGD(learning_rate=0.01, momentum=0.9),
    loss='mse',
    metrics=['mae']
)

train_momentum_ca = model_momentum_ca.fit(
    x_train, y_train,
    epochs=50,
    batch_size=32,
    validation_split=0.2
)
```

```python
model_rmsprop_ca = Sequential([
    Input(shape=(8,)),
    Dense(64, activation='relu'),
    Dense(32, activation='relu'),
    Dense(1)
])

model_rmsprop_ca.compile(
    optimizer=RMSprop(learning_rate=0.001),
    loss='mse',
    metrics=['mae']
)

train_rmsprop_ca = model_rmsprop_ca.fit(
    x_train, y_train,
    epochs=50,
    batch_size=32,
    validation_split=0.2
)
```

```python
model_adam_ca = Sequential([
    Input(shape=(8,)),
    Dense(64, activation='relu'),
    Dense(32, activation='relu'),
    Dense(1)
])

model_adam_ca.compile(
    optimizer=Adam(learning_rate=0.001),
    loss='mse',
    metrics=['mae']
)

train_adam_ca = model_adam_ca.fit(
    x_train, y_train,
    epochs=50,
    batch_size=32,
    validation_split=0.2
)
```

```python
plt.plot(train_sgd_ca.history['mae'], label='SGD')
plt.plot(train_momentum_ca.history['mae'], label='Momentum')
plt.plot(train_rmsprop_ca.history['mae'], label='RMSprop')
plt.plot(train_adam_ca.history['mae'], label='Adam')
plt.xlabel('Epochs')
plt.ylabel('MAE')
plt.title('California Housing - Train MAE')
plt.legend()
plt.show()
```

```python
plt.plot(train_sgd_ca.history['val_mae'], label='SGD')
plt.plot(train_momentum_ca.history['val_mae'], label='Momentum')
plt.plot(train_rmsprop_ca.history['val_mae'], label='RMSprop')
plt.plot(train_adam_ca.history['val_mae'], label='Adam')
plt.xlabel('Epochs')
plt.ylabel('MAE')
plt.title('California Housing - Validation MAE')
plt.legend()
plt.show()
```

```python
scores_ca = {
    'SGD': model_sgd_ca.evaluate(x_test, y_test, verbose=0)[1],
    'Momentum': model_momentum_ca.evaluate(x_test, y_test, verbose=0)[1],
    'RMSprop': model_rmsprop_ca.evaluate(x_test, y_test, verbose=0)[1],
    'Adam': model_adam_ca.evaluate(x_test, y_test, verbose=0)[1]
}
print(scores_ca)

v10_ca = {
    'SGD': train_sgd_ca.history['val_mae'][9],
    'Momentum': train_momentum_ca.history['val_mae'][9],
    'RMSprop': train_rmsprop_ca.history['val_mae'][9],
    'Adam': train_adam_ca.history['val_mae'][9]
}
print('a. Meilleure MAE après 10 epochs :', min(v10_ca, key=v10_ca.get))
print('b. Plus petite erreur sur test :', min(scores_ca, key=scores_ca.get))
```
